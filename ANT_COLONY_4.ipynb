```python

import random

# Distance matrix
dist = [
    [0, 10, 15, 20],
    [10, 0, 35, 25],
    [15, 35, 0, 30],
    [20, 25, 30, 0]
]

n = len(dist)
ants = 10
iterations = 100
alpha = 1
beta = 2
rho = 0.5
Q = 100

pheromone = [[1.0] * n for _ in range(n)]

best_tour = None
best_length = float('inf')

for _ in range(iterations):
    for _ in range(ants):
        tour = [random.randrange(n)]

        while len(tour) < n:
            current = tour[-1]
            unvisited = [i for i in range(n) if i not in tour]

            weights = [
                pheromone[current][j] ** alpha *
                (1 / dist[current][j]) ** beta
                for j in unvisited
            ]

            next_city = random.choices(
                unvisited, weights=weights, k=1
            )[0]
            tour.append(next_city)

        length = sum(
            dist[tour[i]][tour[(i + 1) % n]]
            for i in range(n)
        )

        if length < best_length:
            best_length = length
            best_tour = tour[:]

    # Evaporate pheromones
    for i in range(n):
        for j in range(n):
            pheromone[i][j] *= (1 - rho)

    # Deposit pheromones
    for ant_tour in [best_tour]:
        for i in range(n):
            a = ant_tour[i]
            b = ant_tour[(i + 1) % n]
            pheromone[a][b] += Q / best_length
            pheromone[b][a] += Q / best_length

# Display result
cities = ['A', 'B', 'C', 'D']
route = [cities[i] for i in best_tour]
route.append(route[0])

print("Best Tour:", " -> ".join(route))
print("Shortest Distance:", best_length)
print("Iterations:", iterations)
```
